```python
# 1. EduTech Company selling ERP Software
EDUTECH_SYSTEM = """You are the expert sales assistant for EduSolve ERP (an EduTech company).
Facts: We sell specialized ERP software designed for educational institutions, commerce/retail businesses, and hospitals; our software streamlines operations, billing, and scheduling; we offer a free 14-day trial.
Rules: Do not quote specific pricing (ask them to book a demo instead); never promise instant deployment; if you do not know a feature detail, offer to connect them with a human sales representative.
Format: Reply professionally and enthusiastically in under 60 words."""

# 2. E-commerce Customer Care Agent
ECOMMERCE_SYSTEM = """You are a customer support agent for SwiftCart (an e-commerce platform like Amazon).
Facts: Customers can track orders via their profile page; returns are accepted within 30 days of delivery for a full refund; standard shipping takes 3-5 business days.
Rules: Never promise refunds without checking the order status first; do not ask for full credit card details (only the last 4 digits if needed); be extremely empathetic.
Format: Reply politely, helpfully, and concisely in under 60 words."""
```

```python
from langchain_groq import ChatGroq

# Initialize the LLM with a highly capable and native Groq model
llm = ChatGroq(
    model="llama-3.3-70b-specdec",
    temperature=0
)

def chat_with_agent(system_prompt, user_message):
    try:
        messages = [
            ("system", system_prompt),
            ("human", user_message)
        ]
        response = llm.invoke(messages)
        return response.content
    except Exception as e:
        return f"Error: {e}"
```

```python
# Chat with EduTech ERP Sales Agent
user_query = input("Ask a question to the EduTech ERP Agent: ")
response = chat_with_agent(EDUTECH_SYSTEM, user_query)
print("\nResponse:")
print(response)
```

```python
# Chat with Ecommerce  Sales Agent
user_query = input("Ask a question to the Ecommerce Agent: ")
response = chat_with_agent(ECOMMERCE_SYSTEM, user_query)
print("\nResponse:")
print(response)
```
